```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv("data_for_linear_regression.csv")
```

```python
df.head()
```

```python
#null value check
df.isnull().sum()
```

```python
#delete date column
df=df.drop(columns = ['Date'])
```

```python
df.head()
```

```python
#ei uporer table ta visulize korte nicher code tuku likhte hobe 
# Starting (USD) = x and  Ending (USD) = y , ekhane new ekta x value dekhe y value prediction kora hobe linear regression er maddhome

plt.scatter(df['Starting (USD)'],df['Ending (USD)'])
plt.xlabel('Starting (USD)')
plt.ylabel('Ending (USD)')
plt.title('NASDAQ100 Stock Price')
```

```python
# our target is ending price prediction 
```

```python
x = df.drop(columns = ['Ending (USD)'])
x.head()
```

```python
y = df.drop(columns = ['Starting (USD)'])
y.head()
```

```python
#Linear Regression apply
from sklearn.linear_model import LinearRegression 

```

```python
#function create
reg =  LinearRegression()
```

```python
x.mean()
```

```python
y.mean()
```

```python
plt.scatter(x.mean(),y.mean(),color= 'red')
plt.scatter(df['Starting (USD)'],df['Ending (USD)'])
plt.xlabel('Starting (USD)')
plt.ylabel('Ending (USD)')
plt.title('NASDAQ100 Stock Price')
```

```python
#train the model 
reg.fit(x,y)
```

```python
m = reg.coef_
m
```

```python
c = reg.intercept_
c
```

```python
# predicted result , y= mx+c
m*16700 + c
```

```python
#m,c dhore dhore korar dorkar nai 
reg.predict([[16700]])
```

```python
#df e predict y name e column make
df['predicted_y']= reg.predict(x)
df.head()
```

```python
#plt.plot(x,df['predicted_y'])
plt.plot(x, reg.predict(x))
plt.scatter(x.mean(),y.mean(),color= 'red')
plt.scatter(df['Starting (USD)'],df['Ending (USD)'])
plt.xlabel('Starting (USD)')
plt.ylabel('Ending (USD)')
plt.title('NASDAQ100 Stock Price')

```
